```python
import pandas as pd
import streamlit as st
import numpy as np

```

```python
pedidos = pd.read_csv("KORA_pedidos.csv",parse_dates=["fecha_pedido"])
clientes=pd.read_csv("KORA_clientes.csv",parse_dates=["fecha_registro",])

filtro_canal = st.multiselect(label="CANALES",options= clientes.canal_adquisicion.unique(), default=clientes.canal_adquisicion.unique())
compradores = pedidos.groupby("id_cliente").agg(n_pedidos=("id_pedido","count"))
clientes_FILTRADO= clientes[clientes.canal_adquisicion.isin(filtro_canal)]
final = clientes_FILTRADO.merge(compradores,how="left", on = "id_cliente")
numero_pedidos = final.query("n_pedidos>0").shape[0]
st.metric(label="COMPRADORES",value=numero_pedidos, border= True)
```
